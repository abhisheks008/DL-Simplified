```python
from transformers import pipeline

# Ek dummy text jisko hum summarize karenge
sample_text = """
Artificial intelligence (AI) is intelligence demonstrated by machines, as opposed to the natural intelligence displayed by animals including humans. Leading AI textbooks define the field as the study of "intelligent agents": any system that perceives its environment and takes actions that maximize its chance of achieving its goals. Some popular accounts use the term "artificial intelligence" to describe machines that mimic "cognitive" functions that humans associate with the human mind, such as "learning" and "problem solving". As machines become increasingly capable, tasks considered to require "intelligence" are often removed from the definition of AI, a phenomenon known as the AI effect. A quip in Tesler's Theorem says "AI is whatever hasn't been done yet."
"""

print("Original Text Length:", len(sample_text.split()), "words\n")

# ==========================================
# 1. T5 (Text-to-Text Transfer Transformer)
# ==========================================
print("--- Loading T5 Model ---")
# framework="pt" forces the pipeline to use PyTorch
t5_summarizer = pipeline("summarization", model="t5-small", framework="pt") 

t5_summary = t5_summarizer(sample_text, max_length=50, min_length=15, do_sample=False)
print("\n[T5 Summary]:")
print(t5_summary[0]['summary_text'])
print("-" * 50)


# ==========================================
# 2. BART (Bidirectional and Auto-Regressive Transformers)
# ==========================================
print("--- Loading BART Model ---")
bart_summarizer = pipeline("summarization", model="facebook/bart-large-cnn", framework="pt")

bart_summary = bart_summarizer(sample_text, max_length=50, min_length=15, do_sample=False)
print("\n[BART Summary]:")
print(bart_summary[0]['summary_text'])
print("-" * 50)
```

```python
import nltk
# NLTK ka tokenizer download karna zaroori hai text ko sentences mein todne ke liye
nltk.download('punkt')
nltk.download('punkt_tab')

from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer
from sumy.summarizers.lsa import LsaSummarizer

# Same purana sample text use kar rahe hain
parser = PlaintextParser.from_string(sample_text, Tokenizer("english"))
sentences_count = 2 # Humein summary mein exactly 2 important sentences chahiye

# ==========================================
# 3. TextRank Summarizer (Based on Google's PageRank algorithm)
# ==========================================
print("--- Running TextRank Model ---")
tr_summarizer = TextRankSummarizer()
tr_summary = tr_summarizer(parser.document, sentences_count)

print("\n[TextRank Summary]:")
for sentence in tr_summary:
    print(sentence)
print("-" * 50)

# ==========================================
# 4. LSA Summarizer (Latent Semantic Analysis)
# ==========================================
print("--- Running LSA Model ---")
lsa_summarizer = LsaSummarizer()
lsa_summary = lsa_summarizer(parser.document, sentences_count)

print("\n[LSA Summary]:")
for sentence in lsa_summary:
    print(sentence)
print("-" * 50)
```
